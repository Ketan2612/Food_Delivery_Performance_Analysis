Since rating is on the scale of 1-5 , and 10 persons have rating of 6 so now handling them with average rating value.

```python
avg_rating = df.loc[df["Delivery_person_Ratings"] <= 5, "Delivery_person_Ratings"].mean()
```

```python
df.loc[df["Delivery_person_Ratings"] == 6, "Delivery_person_Ratings"] = avg_rating
```

Target Column

```python
valid_count = pd.to_numeric(df["TARGET"], errors="coerce").notna().sum()
print(valid_count)
```

```python
df["TARGET"] = df["TARGET"].astype(str).str.replace(r"\D", "", regex=True)
```

```python
df.loc[df["TARGET"] == "", "TARGET"] = pd.NA
```

```python
df["TARGET"] = df["TARGET"].str[:4]
```

```python
df["TARGET"] = pd.to_numeric(df["TARGET"], errors="coerce") / 100
```
